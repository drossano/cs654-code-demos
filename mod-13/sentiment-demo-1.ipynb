```python
# sort so positive reviews are at top, negative are at bottom
df.sort_values(by = 'label', ascending= False, inplace= True)
```

```python
df_positive_100 = df.head(100)
```

```python
df_negative_100 = df.tail(100)
```

```python
# convert text to list items
positive_100 = df_positive_100.iloc[:,0].to_list()
negative_100 = df_negative_100.iloc[:,0].to_list()
```

```python

positive_100[:10]
```

```python
from nltk.corpus import stopwords
import string
english_stops = stopwords.words('english')
```

```python
len(english_stops)
```

```python
def my_clean_words(s):
    all_words = []
    review = s.translate(str.maketrans('','',string.punctuation))
    words = review.split()
    for w in words:
        w = w.lower()
        if w not in english_stops:
            all_words.append(w)
    return all_words
```

```python
pos_100_words = []
pos_100_words_list = []

for review in positive_100:
    words_cleaned = my_clean_words(review)
    pos_100_words_list.append(words_cleaned)
    for word in words_cleaned:
        pos_100_words.append(word)
```

```python
print(len(pos_100_words_list))
print(len(pos_100_words))
print(len(set(pos_100_words)))
```

```python
neg_100_words = []
neg_100_words_list = []

for review in negative_100:
    words_cleaned = my_clean_words(review)
    neg_100_words_list.append(words_cleaned)
    for word in words_cleaned:
        neg_100_words.append(word)
```

```python
print(len(neg_100_words_list))
print(len(neg_100_words))
print(len(set(neg_100_words)))
```

```python
print(len(set(neg_100_words+pos_100_words)))
```

```python
attribute_words = list(set(neg_100_words + pos_100_words))
```

```python
# transform pos reviews into one-hot encodings

pos_vectors = []

for i in range(len(pos_100_words_list)):
    vector = []
    for j in range(len(attribute_words)):
        if attribute_words[j] in pos_100_words_list[i]:
            vector.append(1)
        else:
            vector.append(0)
    pos_vectors.append(vector)
```

```python
len(pos_vectors)
```

```python
print(pos_vectors[0])
```

```python
# transform neg reviews into one-hot encodings

neg_vectors = []

for i in range(len(neg_100_words_list)):
    vector = []
    for j in range(len(attribute_words)):
        if attribute_words[j] in neg_100_words_list[i]:
            vector.append(1)
        else:
            vector.append(0)
    neg_vectors.append(vector)
```

```python
len(neg_vectors)
```

```python
len(neg_vectors[0])
```

```python
import numpy as np
x = np.array(pos_vectors + neg_vectors)
```

```python
x.shape
```

```python
pos_negative_labels = []
for i in range(100):
    pos_negative_labels.append(1)
for i in range(100):
    pos_negative_labels.append(0)
```

```python
y = np.array(pos_negative_labels)
```

```python
from sklearn import tree
clf = tree.DecisionTreeClassifier(random_state=42)
clf = clf.fit(x, y)
```

```python
clf.predict(x[:10])
```

```python
clf.predict(x[-10:])
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size= .5, random_state= 42)
```

```python
DT_model = tree.DecisionTreeClassifier(random_state=42)
```

```python
DT_model = DT_model.fit(x_train, y_train)
```

```python
y_train_pred = DT_model.predict(x_train)
y_test_pred = DT_model.predict(x_test)
```

```python
y_test_pred
```

```python
print(accuracy_score(y_true=y_train,y_pred=y_train_pred))
```

```python
print(accuracy_score(y_true=y_test,y_pred=y_test_pred))
```

```python
from sklearn.model_selection import cross_val_score
```

```python
scores = cross_val_score(DT_model, x, y, cv = 2)
```
